```python
import requests

def wikipedia_page(title):
    """
    Returns the raw text of a Wikipedia page given its title.
    """

    url = "https://en.wikipedia.org/w/api.php"

    params = {
        "action": "query",
        "format": "json",
        "titles": title,
        "prop": "extracts",
        "explaintext": True,
        "redirects": 1
    }

    headers = {
        "User-Agent": "WikipediaScraper/1.0 (student project)"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=10
    )

    response.raise_for_status()

    data = response.json()

    page = next(iter(data["query"]["pages"].values()))

    return page.get("extract", "Page not found")


```

```python
text = wikipedia_page("Earth")

print(text[:2000])
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Create WordCloud
wordcloud = WordCloud(
    random_state=0,
    normalize_plurals=False,
    width=600,
    height=300,
    max_words=300,
    stopwords=["the"] # No stop words are removed (e.g., "the", "a", "an", "is", "and", "of", "to")
)
wordcloud.generate(text)
```

```python

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
import requests
import re
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')


# ==========================================
# 1. Get the text
# ==========================================

result = requests.get(
    'http://www.gutenberg.org/files/11/11-0.txt'
)

text = result.text


# ==========================================
# 2. Display Original Text
# ==========================================

print("=" * 70)
print("ORIGINAL TEXT - BEFORE CLEANING")
print("=" * 70)

print(text[:2000])


# ==========================================
# 3. Cleaning using Regular Expressions
# ==========================================

text = text.lower()

text = re.sub(r'[^a-z\s]', '', text)

text = re.sub(r'\s+', ' ', text)

text = text.strip()


# ==========================================
# 4. Display Cleaned Text
# ==========================================

print("\n" + "=" * 70)
print("CLEANED TEXT - AFTER CLEANING")
print("=" * 70)

print(text[:2000])


# ==========================================
# 5. Tokenization
# ==========================================

tokens = word_tokenize(text)


print("\n" + "=" * 70)
print("TOKENS - AFTER TOKENIZATION")
print("=" * 70)

print(tokens[:100])


# ==========================================
# 6. Select Stopwords
# ==========================================

stopwords = {
    'the',
    'of',
    'and',
    'in',
    'to',
    'a',
    'was',
    'is',
    'on',
    'for'
}


# ==========================================
# 7. Remove Selected Stopwords
# ==========================================

clean_tokens = [
    token for token in tokens
    if token not in stopwords
]


# ==========================================
# 8. Display Tokens After Stopword Removal
# ==========================================

print("\n" + "=" * 70)
print("TOKENS - AFTER STOPWORD REMOVAL")
print("=" * 70)

print(clean_tokens[:100])


# ==========================================
# 9. Token Count
# ==========================================

print("\n" + "=" * 70)
print("TOKEN COUNT")
print("=" * 70)

print("Before stopword removal:", len(tokens))
print("After stopword removal:", len(clean_tokens))
```
